```python
# TODO: load_documents("documents.json") and build_embedding_matrix(documents)
# Start in offline mode (the default) while you're building - switch to the real
# API once your pipeline works end to end.
docs = load_documents("documents.json")

matrix = build_embeddings_matrix(docs, mode="offline")
print(f"Built embeddings matrix of shape {matrix.shape}")
```

## Step 3: Run a few example search queries

```python
# 2. Test Query 1: Physics Topic
query1 = "What is the theory of relativity?"
results1 = search(query1, matrix, docs, top_k=3, mode="offline")

print(f"=== Search Results for: '{query1}' ===")
for doc, score in results1:
    print(f"[{score:.4f}] Score | Topic: {doc['topic']} | ID: {doc['id']}")
    print(f"       Text: {doc['text']}")

print("\n" + "="*70 + "\n")

# 3. Test Query 2: Cybersecurity Topic
query2 = "How do hackers trick people into giving away passwords?"
results2 = search(query2, matrix, docs, top_k=3, mode="offline")

print(f"=== Search Results for: '{query2}' ===")
for doc, score in results2:
    print(f"[{score:.4f}] Score | Topic: {doc['topic']} | ID: {doc['id']}")
    print(f"       Text: {doc['text']}")

print("\n" + "="*70 + "\n")

# 4. Test Query 3: History Topic
query3 = "What caused the rise of factories and technological innovation?"
results3 = search(query3, matrix, docs, top_k=3, mode="offline")

print(f"=== Search Results for: '{query3}' ===")
for doc, score in results3:
    print(f"[{score:.4f}] Score | Topic: {doc['topic']} | ID: {doc['id']}")
    print(f"       Text: {doc['text']}")
```
